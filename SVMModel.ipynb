```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
data=pd.read_csv('C:\Clg\TekWorks\Datasets\gender_classification_v7.csv')
df=pd.DataFrame(data)
df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python

df['gender'] = df['gender'].map({'Male':0,'Female':1})
```

```python
x=df.iloc[:,:-1].values
y=df.iloc[:,0].values
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
x_train=scaler.fit_transform(x_train)
x_test=scaler.transform(x_test)
```

Math Function

```python
# eq of hyper plane:


# w.x+b=0

# w= weights

# x=input

# b= bias/intercept

# prediction rule:

# if : w.x+b >= 0 --> Class 1
# else : w.x +b < 0 --> Class 0
```

```python
w=2
x=3
b=-4

y=w*x+b
print(y)

if y>0:
    print("Class 1")
else:
    print("Class 0")
```

```python
from sklearn.svm import SVC
svm_classifier = SVC(kernel='rbf',#Type of Boundary
                     C=1.0,# Regulizer parameter
                     gamma='scale', # Controls the influence of a single training example
                     degree=3
                     )

# train the model 
svm_classifier.fit(x_train,y_train)
# make perdictions
y_pred=svm_classifier.predict(x_test)

from sklearn.metrics import accuracy_score
accuracy_score(y_test,y_pred)
```
